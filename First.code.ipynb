```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

filepath =r"D:/Internship project work/Project 1/CSV File/vadodara_demand_data.csv"   
df = pd.read_csv(filepath, parse_dates = ['timestamp'])

print(df.shape)
print(df.head())
print(df.info())
print(df.tail())
print(df.dtypes)
print(df.describe())
print(df.isnull().sum())
print(df.timestamp.duplicated().sum())

plt.plot(df['timestamp'], df['demand_mw'], color='blue', label='Demand (MW)')
plt.title('Demand over Time')
plt.xlabel('Timestamp')
plt.ylabel('Demand (MW)')
plt.legend()
plt.show()

start_date = pd.Timestamp('2024-01-01 00:00')
end_date = pd.Timestamp('2024-01-08 23:00')
date_range = df.loc[df['timestamp'].between(start_date, end_date)]

plt.plot(date_range['timestamp'], date_range['demand_mw'], color='blue', label='Demand (MW)', linewidth=2)
plt.title('Demand over Time (1st Jan 2024 to 8th Jan 2024)')
plt.xlabel('Timestamp')
plt.ylabel('Demand (MW)')
plt.legend()
plt.show()

copy_df = df.copy()

df = df.set_index('timestamp')
df = df.sort_index()

print("Repeated timestamps before:", df.index.duplicated().sum())
df = df.loc[~df.index.duplicated(keep='first')]
print("Repeated timestamps after:", df.index.duplicated().sum())

hourly_range = pd.date_range(start=df.index.min(), end=df.index.max(), freq='h')
difference_values = hourly_range.difference(df.index)
print("Missing timestamps:", difference_values)

gap_fill = df.reindex(hourly_range)
newdf = gap_fill.interpolate(method='time')

print(newdf.isnull().sum())
print(newdf.shape)
print(newdf['is_holiday'].unique())
print(newdf.duplicated().sum())

save_path = r"D:/Internship project work/Project 1/CSV File/vadodara_demand_data_cleaned.csv"
newdf.to_csv(save_path, index=True)
index_label = df.index.name = 'timestamp'

filepath =r"D:/Internship project work/Project 1/CSV File/vadodara_demand_data_cleaned.csv"   
df = pd.read_csv(filepath)
print(df.shape)
```
